```python
class Song:
    def __init__(self, titol, autor, durada):
        self.titol = titol
        self.autor = autor
        self.durada = durada
    def info(self):
        return f"{self.titol}-{self.autor} ({self.durada} segons)"
class Playlist:
    def __init__(self, song_list):
        self.song_list = song_list
    def afegir_song(self,song):
        self.song_list.append(song)
    def durada_total(self):
        i=0
        for j in self.song_list:
            i+=j.durada
        return f"La durada total de la playlist és de {i} segons"
    def mostrar_playlist(self):
        for k in self.song_list:
            print(k.info())
    def canco_mes_llarga(self):
        i=0
        blud= 0
        for j in self.song_list:
            if j.durada > i:
                i=j.durada
                blud=j.titol
        return f"La canço més llarga de la playlist és {blud} amb {i} segons"     
        
```

```python
song= Song(titol="Flashing Lights",autor="Kanye West", durada=240)
song1=Song(titol="See you again", autor= "Tyler, the Creator", durada=180)
playlist=Playlist([song])
playlist.afegir_song(song1)
playlist.durada_total()
```

```python
playlist.mostrar_playlist()
playlist.canco_mes_llarga()
```
