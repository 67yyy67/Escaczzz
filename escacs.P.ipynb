```python
Tablero = [["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"]
           ]

def mostrar(Tablero):
    print("  a,  b,  c,  d,  e,  f,  g,  h")
    i = 1
    for a in Tablero:
        print(f"{i}{a}")
        i = i + 1

Pieceslist = []
def clear():
    return [["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"],
           ["00", "00", "00", "00", "00", "00", "00", "00"]
           ]
   





class Pawn():
    def __init__(self, x = [0, 0], team = 0):
        self.x = x
        self.team = team
        if self.team == 0:
            self.added_string = "w"
        else:
            self.added_string = "b"
        Pieceslist.append(self)
        Tablero[self.x[0]][self.x[1]] = "p" + self.added_string
        if self.team == 1:
            self.stq = 6
        elif self.team == 0:
            self.stq = 1
    def move(self, z = [0, 0]):
        # EN aquest IF, comprovo si  bla bla bla
        if ((((z[0] == self.x[0] + 1) and (z[1] == self.x[1])) or (((z[1] == self.x[1]) and (z[0] == self.x[0] + 2)) and (self.x[0] == self.stq))) and self.team == 0) or ((((z[0] == self.x[0] - 1) and (z[1] == self.x[1])) or (((z[1] == self.x[1]) and (z[0] == self.x[0] - 2)) and (self.x[0] == self.stq))) and self.team == 1):
            nomou = False
            for a in Pieceslist:
                for b in range(self.x[0], z[0] + 1):
                    if a.x == b:
                        nomou = True
            if nomou == True:
                print("invalid")
            else:
                Tablero[self.x[0]][self.x[1]] = "00"
                self.x = z
                Tablero[z[0]][z[1]] = "p" + self.added_string
        # EWn aquest if, comprvo lab la bla
        elif (((z[0] == self.x[0] + 1) and (z[1] == self.x[1] + 1)) or ((z[1] == self.x[1] - 1) and (z[0] == self.x[0] + 1)) and self.team == 0) or (((z[0] == self.x[0] + 1) and (z[1] == self.x[1] - 1)) or ((z[1] == self.x[1] - 1) and (z[0] == self.x[0] - 1)) and self.team == 1):
            nomou = True
            for a in Pieceslist:
                if a.x == z and a.team != self.team:
                    nomou = False
            if nomou == True:
                print("invalid")
            else:
                Tablero[self.x[0]][self.x[1]] = "00"
                self.x = z
                Tablero[z[0]][z[1]] = "p" + self.added_string
        else:
            print("invalid")

class Knight():
    def __init__(self, x = [0, 0], team = 0):
        self.x = x
        self.team = team
        Pieceslist.append(self)
        if self.team == 0:
            self.added_string = "w"
        else:
            self.added_string = "b"
        Tablero[self.x[0]][self.x[1]] = "N" + self.added_string
    def move(self, z = [0, 0]):
        if (((z[0] == self.x[0] + 1) or (z[0] == self.x[0] - 1)) and ((z[1] == self.x[1] + 2) or (z[1] == self.x[1] - 2))) or (((z[1] == self.x[1] + 2) or (z[1] == self.x[1] - 2)) and ((z[1] == self.x[1] + 1) or (z[1] == self.x[1] - 1))):
            nomou = False
            for a in Pieceslist:
                if a.x == z and a.team == self.team:
                    nomou = True
            if nomou == True:
                print("invalid")
            else:
                Tablero[self.x[0]][self.x[1]] = "00"
                self.x = z
                Tablero[z[0]][z[1]] = "N" + self.added_string
        else:
            print("invalid")

```
